```python
import datetime as dt
import logging

logger = logging.getLogger(__name__)

print("hi here")


async def test_upsert():
    async with cdb.CardDBClient() as client:
        cols = ["cid", "name", "set", "source", "created_ts", "modified_ts"]
        rows = [
            {
                "cid": 2,
                "name": "test",
                "set": "test",
                "source": "test",
                "created_ts": dt.datetime.now(),
                "modified_ts": dt.datetime.now(),
            }
        ]
        await client.upsert_rows(
            "cards",
            cols,
            rows,
            conflict_column="cid",
        )


result = await test_upsert()
result
```

```python
async def test_specialist():
    async with cdb.CardDBClient() as client:
        specialist = cdb.DBSpecialist(client, "cards", cdb.CardRow, debug=True)
        result = await specialist.read_items()
        return result

await test_specialist()
```
